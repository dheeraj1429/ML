```python
from nltk.stem import PorterStemmer, SnowballStemmer, RegexpStemmer
from nltk.tokenize import word_tokenize
from sklearn.feature_extraction.text import TfidfVectorizer
import nltk
```

```python
nltk.download("wordnet")
```

```python
corpus = [
  "Training models on large text datasets",
  "Model evaluation and hyperparameter tuning history",
]
tokens = []
for item in corpus:
  tokens.extend(word_tokenize(item))
tokens
```

```python
porter = PorterStemmer()
snowball = SnowballStemmer("english")
regexp_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)

for word in tokens:
  print(word+'---->', porter.stem(word), "-", snowball.stem(word), "-", regexp_stemmer.stem(word))
```

```python
def stem_tokenizer(text: str) -> list[str]:
  return [snowball.stem(word.lower()) for word in text]

vectorizer = TfidfVectorizer(tokenizer=stem_tokenizer)
X = vectorizer.fit_transform(corpus)
X
```

```python
from nltk.stem import WordNetLemmatizer

"""
POS - Noun - n
Verb - v
Adjective - a
Adverb - r
"""

lemmatizer = WordNetLemmatizer()
for word in tokens:
  print(word+'---->', lemmatizer.lemmatize(word))
```
